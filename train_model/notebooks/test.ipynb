```python
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
```

```python
# 1. Load your best model
model = load_model("resnet_model.h5")
```

```python
# 2. Define a helper to load & preprocess one image
def preprocess_image(path, img_size=(224, 224), to_grayscale=True):
    img = cv2.imread(path)
    if img is None:
        raise FileNotFoundError(f"Cannot read image at {path}")
    img = cv2.resize(img, img_size)
    if to_grayscale:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)[..., np.newaxis]
    else:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = img.astype("float32") / 255.0
    # add batch axis
    return np.expand_dims(img, axis=0)
```

```python
# 3. Predict on a few test images
test_paths = [
    "./chest_xray/test/NORMAL/IM-0017-0001.jpeg",
    "./chest_xray/test/PNEUMONIA/person1_virus_13.jpeg",
]

for p in test_paths:
    x = preprocess_image(p)
    prob = model.predict(x)[0, 0]
    label = "Pneumonia" if prob > 0.5 else "Normal"
    print(f"{p.split('/')[-1]} → {label} (p={prob:.3f})")
```
